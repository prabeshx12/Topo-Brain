## 📊 What to Monitor During Training

**Healthy training signs:**
- `D_loss`: Should stay ~0.3-0.7 (NOT drop to 0.02!)
- `G_loss`: Should gradually decrease
- `Val_PSNR`: Should reach 25+ dB by epoch 50

**If D_loss drops below 0.1:**
- Training may be collapsing
- Consider stopping and further reducing D learning rate
